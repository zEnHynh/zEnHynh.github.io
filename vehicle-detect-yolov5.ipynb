```python
import os
import zipfile

def unzip_all(directory):
    """
    Hàm này sẽ tìm và giải nén tất cả các file ZIP trong thư mục chỉ định và các thư mục con.
    """
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.endswith('.zip'):
                file_path = os.path.join(root, file)
                extract_path = os.path.splitext(file_path)[0]  # Tạo thư mục giải nén cùng tên
                os.makedirs(extract_path, exist_ok=True)

                with zipfile.ZipFile(file_path, 'r') as zip_ref:
                    zip_ref.extractall(extract_path)

                print(f"Đã giải nén {file_path} vào {extract_path}")

                # Sau khi giải nén, có thể xóa file zip nếu không cần thiết
                os.remove(file_path)
                print(f"Đã xóa file ZIP: {file_path}")

                # Gọi đệ quy để giải nén các file ZIP bên trong nếu có
                unzip_all(extract_path)

# Thư mục chứa các file ZIP đã giải nén
labels_dir = '/content/DOTA/labels/train'

# Gọi hàm để giải nén tất cả các file ZIP đệ quy
unzip_all(labels_dir)

print("Giải nén tất cả các file ZIP đệ quy hoàn tất.")
```

```python
from PIL import Image
import os

# Định nghĩa các lớp mục tiêu
target_classes = ['car', 'truck', 'ship', 'other']

# Tạo mapping từ các lớp DOTA sang các lớp mục tiêu
class_mapping_dota_to_target = {
    'small-vehicle': 'car',
    'large-vehicle': 'truck',
    'ship': 'ship',
    # Các lớp khác sẽ được map sang 'other'
}

# Tạo mapping từ lớp mục tiêu sang class_id
class_to_id = {cls: idx for idx, cls in enumerate(target_classes)}

def convert_dota_labels_to_yolo(input_label_dir, output_label_dir, images_dir, class_mapping, target_classes, filter_difficulty=None):
    """
    Args:
        input_label_dir (str): Đường dẫn đến thư mục chứa file nhãn gốc.
        output_label_dir (str): Đường dẫn đến thư mục sẽ lưu file nhãn đã chuyển đổi.
        images_dir (str): Đường dẫn đến thư mục chứa ảnh.
        class_mapping (dict): Mapping từ class_name DOTA sang class_name mục tiêu.
        target_classes (list): Danh sách các lớp mục tiêu.
        filter_difficulty (list, optional): Danh sách các mức độ khó để lọc. Nếu None, không lọc.
    """
    if not os.path.exists(output_label_dir):
        os.makedirs(output_label_dir)

    for filename in os.listdir(input_label_dir):
        if not filename.endswith('.txt'):
            continue
        # Tìm file ảnh tương ứng (jpg hoặc png)
        image_filename_jpg = filename.replace('.txt', '.jpg')
        image_filename_png = filename.replace('.txt', '.png')
        image_path_jpg = os.path.join(images_dir, image_filename_jpg)
        image_path_png = os.path.join(images_dir, image_filename_png)

        if os.path.exists(image_path_jpg):
            image_path = image_path_jpg
        elif os.path.exists(image_path_png):
            image_path = image_path_png
        else:
            print(f"Không tìm thấy ảnh tương ứng cho {filename}")
            continue

        # Lấy kích thước ảnh
        with Image.open(image_path) as img:
            img_width, img_height = img.size

        # Đọc file nhãn gốc
        with open(os.path.join(input_label_dir, filename), 'r') as f:
            lines = f.readlines()

        yolo_labels = []
        for line in lines:
            parts = line.strip().split()
            if len(parts) < 10:
                continue  # Bỏ qua nếu định dạng không đúng
            # Lấy 8 số đầu tiên là bốn cặp (x, y)
            coords = list(map(float, parts[:8]))
            class_name_dota = parts[8]
            difficulty = int(parts[9])

            # Nếu filter_difficulty được chỉ định, lọc theo difficulty
            if filter_difficulty is not None and difficulty not in filter_difficulty:
                continue

            # Áp dụng mapping từ class DOTA sang class mục tiêu
            class_name_target = class_mapping.get(class_name_dota, 'other')
            if class_name_target not in target_classes:
                class_name_target = 'other'

            # Lấy class_id từ class_to_id
            class_id = class_to_id.get(class_name_target, class_to_id['other'])

            # Chuyển đổi bbox sang định dạng YOLO
            x_coords = coords[0::2]
            y_coords = coords[1::2]

            x_min = min(x_coords)
            x_max = max(x_coords)
            y_min = min(y_coords)
            y_max = max(y_coords)

            x_center = (x_min + x_max) / 2 / img_width
            y_center = (y_min + y_max) / 2 / img_height
            width = (x_max - x_min) / img_width
            height = (y_max - y_min) / img_height

            # Đảm bảo các giá trị nằm trong khoảng [0,1]
            x_center = min(max(x_center, 0), 1)
            y_center = min(max(y_center, 0), 1)
            width = min(max(width, 0), 1)
            height = min(max(height, 0), 1)

            yolo_labels.append(f"{class_id} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}\n")

        # Ghi các nhãn đã chuyển đổi vào file mới
        with open(os.path.join(output_label_dir, filename), 'w') as f:
            f.writelines(yolo_labels)

# Định nghĩa đường dẫn và mapping lớp
input_label_train = '/content/yolov5_dataset/labels/train'
output_label_train_yolo = '/content/yolov5_dataset/label/train_yolo'

input_label_val = '/content/yolov5_dataset/labels/val'
output_label_val_yolo = '/content/yolov5_dataset/label/val_yolo'

images_train_dir = '/content/yolov5_dataset/images/train'
images_val_dir = '/content/yolov5_dataset/images/val'

# Chuyển đổi nhãn cho tập train với tất cả các mức độ khó
convert_dota_labels_to_yolo(
    input_label_dir=input_label_train,
    output_label_dir=output_label_train_yolo,
    images_dir=images_train_dir,
    class_mapping=class_mapping_dota_to_target,
    target_classes=target_classes,
    filter_difficulty=[0, 1]  # Bạn có thể thay đổi thành [0,1,2] để lọc theo difficulty
)

# Chuyển đổi nhãn cho tập val với tất cả các mức độ khó
convert_dota_labels_to_yolo(
    input_label_dir=input_label_val,
    output_label_dir=output_label_val_yolo,
    images_dir=images_val_dir,
    class_mapping=class_mapping_dota_to_target,
    target_classes=target_classes,
    filter_difficulty=[0, 1]
)
```

```python
import os

# Đường dẫn đến thư mục chứa các tệp nhãn
label_path = '/content/yolov5_dataset/labels/train/'

# Lặp qua tất cả các tệp nhãn
for label_file in os.listdir(label_path):
    if label_file.endswith('.txt'):
        # Đọc nội dung tệp nhãn
        with open(os.path.join(label_path, label_file), 'r') as file:
            lines = file.readlines()

        # Loại bỏ các dòng chứa 'imagesource' và 'gsd' nhưng giữ lại các dòng khác
        filtered_lines = [line for line in lines if not line.startswith('imagesource') and not line.startswith('gsd')]

        # Nếu còn lại dữ liệu sau khi lọc, ghi lại vào tệp nhãn
        if filtered_lines:
            with open(os.path.join(label_path, label_file), 'w') as file:
                file.writelines(filtered_lines)
        else:
            print(f"Tệp {label_file} không còn nội dung nào sau khi lọc.")
```
